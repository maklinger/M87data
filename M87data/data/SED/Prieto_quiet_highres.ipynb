```python

# (frequency_Hz, flux_Jy, error_Jy, angular_scale_arcsec, instrument_label)
raw = [
    (2.06e15,  4.14e-5,  4.1e-6,   0.12*2, "STIS-F25SRF2"),
    (1.36e15,  1.08e-4,  0.02e-4,  0.13*2, "ACS-HRC F220W"),
    (1.27e15,  1.81e-4,  2.e-6,    0.12*2, "STIS-F25QTZ"),
    (1.10e15,  1.28e-4,  0.01e-4,  0.13*2, "ACS-HRC F250W"),
    (8.93e14,  1.79e-4,  1.4e-5,   0.13*2, "ACS-HRC F330W"),
    (6.32e14,  4.13e-4,  0.54e-4,  0.13*2, "ACS-HRC F475W"),
    (4.99e14,  4.24e-4,  0.24e-4,  0.13*2, "ACS-HRC F606W"),
    (3.70e14,  6.34e-4,  1.26e-4,  0.15*2, "ACS-HRC F814W"),
    (3.32e14,  1.30e-3,  0.14e-3,  0.14*2, "ACS-WFC F850LP"),
    (2.47e14,  1.61e-3,  0.16e-3,  0.15*2, "NIC2 F110W"),
    (2.3e11,   0.98,     0.04,     0.040,  "VLBI 1.3mm (Doeleman et al. 2012)"),   # FWHM = 40mas
    (1.00e11,  0.50,     0.1,      np.nan, "VLBI 3mm (Lonsdale et al. 1998)"),
    (86.0e9,   0.16,     0.07,     100e-6, "VLBI 3.5mm (Lee et al. 2008)"),        # beam 198x78 uas -> not a simple radius
    (22.0e9,   0.35,     0.06,     5e-4, "VLBI (Junor & Biretta 1995)"),         # beam 1.15x0.14 mas
    (8.4e9,    0.52,     0.05,     1e-3, "VLBI (Morabito et al. 1986)"),         # beam < 1mas
    (2.3e9,    0.63,     0.08,     3e-3, "VLBI (Morabito et al. 1988)"),         # beam < 3mas
    (1.6e9,    0.364,    np.nan,   4e-3, "VLBI (Giovannini et al. 1990)"),       # beam 4x4 mas
]

rows = []
for freq, flux, err, scale, instrument in raw:
    nuFnu = freq * flux * 1e-23 * 1e12
    sigma = freq * err  * 1e-23 * 1e12 if not np.isnan(err) else np.nan
    rows.append([instrument, freq, scale, nuFnu, sigma])

df = pd.DataFrame(rows, columns=[
    "Instrument",
    "Frequency_Hz",
    "Angular_Scale_arcsec",
    "nuFnu_1e-12ergscm2",
    "sigma_nuFnu_1e-12ergscm2"
])

df.to_csv("M87SED_Prieto_quiet_highres.csv", index=False)
```
