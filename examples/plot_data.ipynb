```python
# Lucchini++16
# 0.5 - 9keV
data_Lucc16 = [
    [-2.04, 0.05, 5.1e-4, 3e-5, 1],
]
# diff. flux: dN/dE = A * (E/E0)^gamma
def butterfly_dif(E, gamma, siggamma, A, sigA, E0):
    EFE = A * E0**2 * (E/E0)**(gamma+2)
    
    dEFEdA = EFE / A
    dEFEdgamma = EFE * np.log(E/E0)
    sigEFE = np.sqrt((dEFEdA * sigA)**2 + (dEFEdgamma*siggamma)**2)
    return EFE, sigEFE
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax, norm = add_SED(
    ax, dataset="M87SED_HST2003", colorbar=False, cmap=colmap,
    angular_scale_color=True, marker="s", #fixed_color="k",
    absorbed=False, frequencymarkers=False, label="2003 HST")
ax, norm = add_SED(
    ax, dataset="M87SED_JWST2022_av300", colorbar=False, cmap=colmap,
    angular_scale_color=True, marker="p", #fixed_color="k",
    absorbed=False, frequencymarkers=False, label="2022 JWST")
ax, norm = add_SED(
    ax, dataset="M87SED_Prieto_quiet_IR", colorbar=False, cmap=colmap,
    angular_scale_color=True, marker="d", #fixed_color="k",
    absorbed=False, frequencymarkers=False, label="1998-2001 Keck/Gemini")
ax, norm = add_SED(
    ax, dataset="M87SED_ALMA2012", colorbar=False, cmap=colmap,
    angular_scale_color=True, marker="x",
    absorbed=False, frequencymarkers=False, label="2012 ALMA")
ax, norm = add_SED(
    ax, dataset="M87SED_Prieto_quiet_radio", colorbar=False, cmap=colmap,
    angular_scale_color=True, marker="*", #fixed_color="k",
    absorbed=False, frequencymarkers=False, label="1999-2005 VLA")
ax, norm = add_SED(
    ax, dataset="M87SED_EHTMWL2017", colorbar=False, cmap=colmap,
    angular_scale_color=True, marker="^", #fixed_color="k",
    absorbed=False, frequencymarkers=False, label="2017 EHT MWL")
ax, norm = add_SED(
    ax, dataset="M87SED_EHTMWL2018", colorbar=True, cmap=colmap,
    angular_scale_color=True, marker=".", #fixed_color="r",
    absorbed=False, frequencymarkers=True, label="2018 EHT MWL",
    ymin=1e-14, ymax=1e-10)


Es_Chandra = np.logspace(np.log10(0.5), np.log10(9), 30) 
for gamma, siggamma, A, sigA, E0 in data_Lucc16:
    EFE, sigEFE = butterfly_dif(Es_Chandra, gamma, siggamma, A, sigA, E0)
    print(gamma, siggamma, A, sigA, E0)
    rChandra = 3e5 
    col = colmap(norm(np.log10(rChandra)))
    # col="k"
    ax.fill_between(Es_Chandra*1e3, (EFE-sigEFE)*c.eV2erg*1e3, (EFE+sigEFE)*c.eV2erg*1e3,
                     color=col,edgecolor="k", alpha=0.3, zorder=100, label="2012 Chandra")
    ax.loglog(Es_Chandra*1e3, EFE*c.eV2erg*1e3, color=col, zorder=100, lw=1)

ax.legend(loc="lower center", ncol=2)
ax.set_xlim(3e-6, 1e5)
ax.set_xticks(10**np.arange(-5., 5.1, 0.5))
# ax.set_ylim(1e-14, 1e-11)
# ax.set_title("radio and ALMA")
fig.tight_layout()
```
