```python
import matplotlib.pyplot as plt
%matplotlib widget
from M87data.threeMLPlugins.M87_2018_Plugin import M87_2018_Plugin
from M87data.threeMLPlugins.M87_2017_Plugin import M87_2017_Plugin
from M87data.threeMLPlugins.M87_Prieto_Plugin import M87_Prieto_Plugin
from M87data.threeMLPlugins.M87_JWST_Plugin import M87_JWST_Plugin
```

```python
djwst = M87_JWST_Plugin("jswt", 1000)
djwst_shifted = M87_JWST_Plugin("jswt", 1000)
djwst_shifted.use_effective_area_correction(0.8, 11)
djwst_shifted.fix_effective_area_correction(3)
```

```python
dradio = M87_2018_Plugin("radio", "radio")
douv = M87_2018_Plugin("ouv", "optical")
dxray = M87_2018_Plugin("xray", "xray")
dgev = M87_2018_Plugin("gev", "gev")
dtev = M87_2018_Plugin("tev", "tev")
```

```python
%matplotlib widget
fig, ax = plt.subplots()

ax = dradio.add2ax_SED_eV_ergscm2(ax)
ax = douv.add2ax_SED_eV_ergscm2(ax)
ax = dxray.add2ax_SED_eV_ergscm2(ax)
ax = dgev.add2ax_SED_eV_ergscm2(ax)
ax = dtev.add2ax_SED_eV_ergscm2(ax)

ax = djwst.add2ax_SED_eV_ergscm2(ax, plot_effective_area_correction=True)
ax = djwst_shifted.add2ax_SED_eV_ergscm2(ax, plot_effective_area_correction=True, color="r")

ax.set_xscale("log")
ax.set_yscale("log")
```
